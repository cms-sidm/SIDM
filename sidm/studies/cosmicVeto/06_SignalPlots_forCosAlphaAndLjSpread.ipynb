```python
import os, sys
sidm_path = str(os.getcwd()).split('/sidm')[0]
if sidm_path not in sys.path:
    sys.path.insert(1, sidm_path)
from sidm.tools import utilities, scaleout, sidm_processor, llpnanoaodschema
from sidm.studies.cosmicVeto import functions

import matplotlib.pyplot as plt
import mplhep as hep
utilities.set_plot_style()
```

```python
channel_name = 'base_ljDisplacementIso'

output_signal_2mu = functions.load_output("signal_2mu2e", channel_name)
output_signal_4mu = functions.load_output("signal_4mu", channel_name)

S = [
    '2Mu2E_1000GeV_0p25GeV_0p002mm',
    '2Mu2E_1000GeV_0p25GeV_2p0mm',
    '4Mu_1000GeV_0p25GeV_0p002mm',
    '4Mu_1000GeV_0p25GeV_2p0mm',
]


histogram_list = [

"cosAlpha_mu", 
    "min_cosAlpha_mu"
]
colors = ["blue", "orange", "red", "green", "purple",]
for i, histogram_name in enumerate(histogram_list):
    plt.figure(figsize=(15, 12))
    for j,s in enumerate(S):
        label = functions.get_signal_label(s)
        if label[0]=="2":
            histogram= output_signal_2mu["out"][s]["hists"][histogram_name][channel_name, ::3j]
        elif label[0] == "4":
            histogram= output_signal_4mu["out"][s]["hists"][histogram_name][channel_name, ::3j]
        else:
            print(f"output not found for{s}")
        utilities.plot(histogram, label=label, linewidth=2, color =colors[j], density=True)
   
    plt.legend(ncols = 2)
    plt.yscale("log")
    plt.ylim(1e-4, 1e3)
    plt.xticks(rotation=90)
    plt.savefig(f"Plots/{channel_name}_{histogram_name}_signal.png", dpi=150)
    plt.show()
    plt.close()
```
