```python
import os, sys
import matplotlib.pyplot as plt


sidm_path = str(os.getcwd()).split('/sidm')[0]
if sidm_path not in sys.path:
    sys.path.insert(1, sidm_path)

from sidm.tools import utilities
from sidm.studies.cosmicVeto import functions
utilities.set_plot_style()
%matplotlib inline
```

```python

channel_name = "cosmic_muons"
data = [ 
    "DoubleMuon_2018C",
]
folder_name = f"OutputFiles/{channel_name}"
sum_data_all= utilities.sum_hist(data, folder_name)
```

```python



    
    
histogram_name = "dsaMuon_eta_phi"
hist_2d =  sum_data_all[histogram_name][channel_name, :, :]
print(hist_2d.axes)

plt.figure()
hist_eta = functions.plot_1d(hist_2d, "dsaMuons_eta")

utilities.plot(hist_eta)
plt.yscale("log")
plt.figure()
hist_phi = functions.plot_1d(hist_2d, "dsaMuons_phi")

utilities.plot(hist_phi)
plt.yscale("log")


histogram_name = "muon_eta_phi"
hist_2d =  sum_data_all[histogram_name][channel_name, :, :]
print(hist_2d.axes)

plt.figure()
hist_eta = functions.plot_1d(hist_2d, "muons_eta")

utilities.plot(hist_eta)
# plt.yscale("log")
plt.figure()
hist_phi = functions.plot_1d(hist_2d, "muons_phi")

utilities.plot(hist_phi)
# plt.yscale("log")
```
